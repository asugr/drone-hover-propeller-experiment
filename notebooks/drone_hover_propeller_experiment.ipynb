# 🚁 Why Does Propeller Size Matter for Drones?

## A simple computational experiment in rotor physics

**Research question:** How does propeller diameter affect the ideal power required for a drone to hover?

This project uses a simplified momentum-theory model to connect drone mass, rotor disk area, thrust, induced air velocity, and ideal hover power.

## 1. Start with the engineering requirement

A hovering drone must produce upward thrust approximately equal to its weight:

$$T = mg$$

where $m$ is drone mass and $g$ is gravitational acceleration.

For a multi-rotor drone, the total thrust is shared among the rotors.

## 2. Why rotor disk area matters

A simplified ideal momentum-theory relationship for a rotor in hover is:

$$v_i = \\sqrt{\\frac{T}{2\\rho A}}$$

where $A$ is total rotor disk area and $\\rho$ is air density.

The corresponding ideal induced power is approximately:

$$P_i = T v_i$$

This model predicts that increasing rotor disk area reduces the induced velocity needed to generate the same thrust.

```python
import numpy as np
import matplotlib.pyplot as plt

rho = 1.225
g = 9.81
mass = 2.0
number_of_rotors = 4
diameters = np.linspace(0.15, 0.40, 100)

T = mass * g
A_total = number_of_rotors * np.pi * (diameters/2)**2
vi = np.sqrt(T / (2*rho*A_total))
P = T * vi

plt.plot(diameters*100, P)
plt.xlabel('Propeller diameter (cm)')
plt.ylabel('Ideal induced hover power (W)')
plt.title('Effect of propeller diameter on ideal hover power')
plt.show()
```

## 3. The experiment

I vary propeller diameter while keeping drone mass and the number of rotors constant.

The model predicts that larger propellers create a larger total rotor disk area. The same required thrust can therefore be generated with a lower induced velocity.

## 4. Important limitation

This is an **idealized hover model**. It does not represent the complete electrical power required by a real drone.

Real drones also lose energy through blade profile drag, motor and ESC losses, turbulence, rotor-rotor interactions, and other aerodynamic effects.

The model is therefore useful for understanding the physics, not for selecting a real drone motor or propeller.

## 5. Engineering conclusion

For the same required thrust, increasing total rotor disk area can reduce ideal induced hover power.

This helps explain an important drone-design trade-off: larger propellers can improve hover efficiency, but they also affect vehicle size, mass, response, packaging, and operating speed.
